```python
import yaml
import subprocess

# Change directory only if not already there
folder = "./syxplain"
if os.getcwd() != os.path.abspath(folder):
    os.chdir(folder)
    print(f"Changed directory to: {os.getcwd()}")
else:
    print(f"Already in folder: {os.getcwd()}")

# Install dependecies (repo has a conda env config)
with open('./syxplain_environment.yml', 'r') as file:
    env_data = yaml.safe_load(file)

# Extract the dependencies from the YAML file
dependencies = env_data.get('dependencies', [])

# Install the dependencies via pip
for dep in dependencies:
    # If the dependency contains a version specifier (e.g., 'numpy>=1.21')
    if '>=' in dep:
        package, version = dep.split('>=')
        subprocess.run(['pip', 'install', '-q', f'{package}>={version}'])
    else:
        # If no version is specified, install the package normally
        subprocess.run(['pip', 'install', '-q', dep])

# Check if PyTorch is listed and install it if necessary
if 'pytorch' in [dep.lower() for dep in dependencies]:
    subprocess.run(['pip', 'install', '-q', 'torch>=2.0'])

# Notify the user when installation is complete
print("All dependencies from env.yml have been installed.")
```

```python
import json
import pandas as pd
from syxplain_lib import State, run_syxplain
```

## Run example 1

```python
def entry_point(state: State) -> State:
    # Load research problem description
    with open("./data/example_1/problem_description.json", "r") as f:
        problem_description = json.load(f)
    problem_description_src = json.dumps(problem_description, indent=4)
    problem_description = json.loads(problem_description_src)
    state["problem_description_src"] = problem_description_src
    state["target_column"] = problem_description["target_variable"]
    state["predictor_columns"] = list(problem_description["features"].keys())

    # Load data
    sample_data = pd.read_csv("./data/example_1/sample_data.csv")
    state["data"] = sample_data

    return state
```

```python
run_syxplain(entry_point)
```
