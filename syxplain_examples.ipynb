## Run example 3

```python
def entry_point(state: State) -> State:
    # Load research problem description
    with open("./data/example_3/problem_description.json", "r") as f:
        problem_description = json.load(f)
    problem_description_src = json.dumps(problem_description, indent=4)
    problem_description = json.loads(problem_description_src)
    state["problem_description_src"] = problem_description_src
    state["target_column"] = problem_description["target_variable"]
    state["predictor_columns"] = list(problem_description["features"].keys())

    # Load data
    datasets = {}
    filenames = [
        "Celegans_AggregatedData.csv",
        "Larva_LocalDensity_AggregatedData.csv",
        "Hemibrain_LocalDensity_AggregatedData.csv",
        "MANC_LocalDensity_AggregatedData.csv",
        "FlyWire_LocalDensity_AggregatedData.csv",
        "Zebrafish_LocalDensity_AggregatedData.csv",
        "MM3_LocalDensity_AggregatedData.csv",
        "H01_LocalDensity_AggregatedData.csv"
    ]

    # Clean species names for display
    species_names = {
        "Celegans_AggregatedData.csv": "C. elegans",
        "Larva_LocalDensity_AggregatedData.csv": "Fly Larva",
        "Hemibrain_LocalDensity_AggregatedData.csv": "Fly Hemibrain",
        "MANC_LocalDensity_AggregatedData.csv": "Fly MANC",
        "FlyWire_LocalDensity_AggregatedData.csv": "Fly FlyWire",
        "Zebrafish_LocalDensity_AggregatedData.csv": "Zebrafish",
        "MM3_LocalDensity_AggregatedData.csv": "Mouse MM3",
        "H01_LocalDensity_AggregatedData.csv": "Human H01"
    }

    print("Loading datasets...")
    all_data = []
    for filename in filenames:
        try:
            # Load
            df = pd.read_csv(f"./data/example_3/{filename}")
            datasets[filename] = df
            #print(f"Loaded {filename}: {df.shape[0]} records, {df.shape[1]} columns")
            #print(df.isna().sum())

            # Normalize datasets for common scale (Kosmos normalization)
            median = df[state["target_column"]].median()
            std = df[state["target_column"]].std()
            df[state["target_column"]] = (df[state["target_column"]] - median) / std

            for col in state["predictor_columns"]:
                median = df[col].median()
                std = df[col].std()
                df[col] = (df[col] - median) / std

            all_data.append(df)
        except Exception as e:
            print(f"Error loading {filename}: {e}")
    print(f"\nSuccessfully loaded {len(datasets)} datasets")

    all_data = pd.concat(all_data, ignore_index=True)

    print(f"All data: {all_data.shape[0]} records, {all_data.shape[1]} columns")
    all_data = all_data.drop_duplicates()
    all_data = all_data.dropna()
    print(f"All data after cleanin: {all_data.shape[0]} records, {all_data.shape[1]} columns")

    state["data"] = all_data

    return state
```

```python
run_syxplain(entry_point)
```
